이번 문제에서는 제목과 설명을 표시하고, 사용자가 `Bike_Sharing_Demand.csv`를 직접 업로드하도록 만듭니다.

2. `st.title()`을 사용해 `"Bike Sharing 시계열 예측"`이라는 제목을 표시합니다.
3. `st.write()`를 사용해 `"월별 자전거 대여량을 확인하고 미래 값을 예측합니다."`라는 설명을 표시합니다.

# 필수 2. 업로드한 데이터를 시계열 데이터로 정리하고 예측하기

## 배경

업로드한 Bike Sharing Demand 원본 데이터는 시간 단위 관측값을 포함하고 있습니다.

웹앱에서는 `datetime`을 날짜/시간으로 변환하고, 월 단위로 전체 대여량을 합계하여 시계열 분석용 데이터를 만듭니다.

## 문제 1. 월별 시계열 데이터를 만드세요.

### 요구사항

파일이 업로드된 경우 다음 작업을 수행합니다.

1. `datetime`을 `pd.to_datetime()`으로 변환합니다.
2. 날짜 순서대로 정렬합니다.
3. `datetime`에서 월 정보를 추출합니다.
4. 월별 `count` 합계를 계산합니다.
5. 월을 인덱스로 설정합니다.
6. `st.dataframe()`으로 월별 데이터를 표시합니다.
7. `st.line_chart()`로 월별 대여량 변화를 표시합니다.

### 확인 포인트

- 시간 순서가 유지되는가?
- 한 행이 한 달을 나타내는 데이터로 정리되었는가?
- 표와 선 그래프가 모두 표시되는가?

````python
if uploaded_file is not None:
    bike['datetime'] = pd.to_datetime(bike['datetime'])
    bike = bike.sort_values('datetime').reset_index(drop=True)
    bike['month'] = bike['datetime'].dt.to_period('M').dt.to_timestamp()
    monthly = bike.groupby('month')[['count']].sum().asfreq('MS')
    st.dataframe(monthly)
    st.line_chart(monthly['count'])
````

## 문제 2. 사용자가 예측 기간을 선택하도록 하세요.

### 요구사항

1. `st.selectbox()`를 사용합니다.
2. 사용자가 `[1, 3, 6]`개월 중 하나를 선택할 수 있도록 합니다.
3. 선택한 값을 `forecast_period` 변수에 저장합니다.

### 확인 포인트

- 사용자 입력이 Python 변수에 저장되는가?
- 선택값을 이후 `forecast()`의 `steps`에 사용할 수 있는가?

````python
forecast_period = st.selectbox('예측 기간(개월)',[1,3,6])
````

### Q2. 사용자가 `3`을 선택했다면 `forecast_period`에는 어떤 값이 저장되나요?

**답변:**  
정수 3이 저장됩니다. 이후 forecast(steps=forecast_period)는 다음 3개월의 값을 예측합니다.

## 문제 3. ARIMA 모델로 미래 값을 예측하세요.

### 요구사항

1. `ARIMA`를 불러옵니다.
2. 월별 `count` 전체 데이터를 사용합니다.
3. 교안과 동일하게 `order=(2, 1, 1)`을 사용합니다.
4. `fit()`으로 모델을 학습합니다.
5. `forecast(steps=forecast_period)`로 선택한 기간만큼 예측합니다.

### 확인 포인트

- 이번 웹앱에서는 현재까지의 전체 데이터를 학습에 사용했는가?
- Test 데이터와 비교하는 방식이 아니라 아직 관측되지 않은 미래를 예측한다는 점을 이해했는가?

````python
from statsmodels.tsa.arima.model import ARIMA
if uploaded_file is not None:
    model = ARIMA(monthly['count'],order=(2,1,1))
    model_fit = model.fit()
    forecast = model_fit.forecast(steps=forecast_period)
````

# 과제 1. 시계열 데이터 예측 웹앱 완성하기

## 배경

필수 실습에서 Streamlit의 기본 화면, CSV 업로드, 월별 시계열 데이터 정리, 예측 기간 선택, ARIMA 예측까지 연습했습니다.

이번 과제에서는 앞에서 만든 기능을 하나의 `app.py`에 연결합니다.

> 새로운 기법을 추가하지 않고 필수 실습에서 사용한 기능을 연결하는 수준입니다.

## 요구사항

1. `"Bike Sharing 시계열 예측"` 제목을 표시합니다.
2. `Bike_Sharing_Demand.csv`를 업로드할 수 있도록 합니다.
3. 업로드한 데이터를 pandas DataFrame으로 불러옵니다.
4. `datetime`을 날짜/시간 자료형으로 변환합니다.
5. 시간 순서대로 정렬한 뒤 월별 `count` 합계를 계산합니다.
6. 월별 데이터를 표와 선 그래프로 표시합니다.
7. 사용자가 `1`, `3`, `6`개월 중 예측 기간을 선택합니다.
8. 전체 월별 `count`로 `ARIMA(2, 1, 1)`을 학습합니다.
9. 선택한 기간만큼 미래 값을 예측합니다.
10. 마지막 관측 월의 다음 달부터 미래 날짜를 생성합니다.
11. 미래 날짜와 예측값을 DataFrame으로 만들어 표로 표시합니다.
12. 기존 실제값과 미래 예측값을 하나의 선 그래프로 표시합니다.

### 확인 포인트

- 파일 업로드 → 데이터 처리 → 사용자 입력 → ARIMA 예측 → 결과 출력 흐름이 연결되는가?
- 미래 날짜의 개수가 사용자가 선택한 예측 기간과 같은가?
- 기존 실제값과 미래 예측값을 구분하여 표시하는가?
- 미래 실제값이 아직 없으므로 MAE와 RMSE를 계산하지 않았는가?

````python
from pathlib import Path

# 아래 전체 코드를 app.py로 저장합니다.
app_code = '''
import pandas as pd
import streamlit as st
from statsmodels.tsa.arima.model import ARIMA

st.title('Bike Sharing 시계열 예측')
st.write('월별 자전거 대여량을 확인하고 미래 값을 예측합니다.')
uploaded_file = st.file_uploader('Bike_Sharing_Demand.csv 업로드', type=['csv'])
if uploaded_file is not None:
    try:
        bike = pd.read_csv(uploaded_file)
        if not {'datetime', 'count'}.issubset(bike.columns):
            st.error('datetime과 count 컬럼이 필요합니다.')
            st.stop()
        st.write('원본 데이터')
        st.dataframe(bike)
        bike['datetime'] = pd.to_datetime(bike['datetime'], errors='raise')
        bike['count'] = pd.to_numeric(bike['count'], errors='raise')
        if bike[['datetime', 'count']].isna().any().any():
            st.error('datetime과 count의 결측치를 확인하세요.')
            st.stop()
        bike = bike.sort_values('datetime').reset_index(drop=True)
        bike['month'] = bike['datetime'].dt.to_period('M').dt.to_timestamp()
        monthly = bike.groupby('month')[['count']].sum().asfreq('MS')
        if len(monthly) < 6 or monthly['count'].isna().any():
            st.error('최소 6개월의 연속 월별 관측 데이터가 필요합니다. 누락 월을 확인하세요.')
            st.stop()
        st.write('월별 관측 대여량 합계')
        st.dataframe(monthly)
        st.line_chart(monthly['count'])
        st.write('이 CSV의 합계는 관측된 날짜 기준입니다. 매월 일부 날짜만 있으면 전체 월의 총수요가 아닙니다.')
        forecast_period = st.selectbox('예측 기간(개월)', [1, 3, 6])
        model_fit = ARIMA(monthly['count'], order=(2, 1, 1)).fit()
        forecast = model_fit.forecast(steps=forecast_period)
        future_dates = pd.date_range(start=monthly.index[-1] + pd.offsets.MonthBegin(1),
                                     periods=forecast_period, freq='MS')
        forecast_df = pd.DataFrame({'predicted': forecast.to_numpy()}, index=future_dates)
        forecast_df.index.name = 'month'
        st.write('미래 예측값')
        st.dataframe(forecast_df)
        chart_df = pd.concat([monthly.rename(columns={'count': 'actual'}), forecast_df], axis=1)
        st.line_chart(chart_df)
    except (ValueError, KeyError, pd.errors.ParserError) as error:
        st.error(f'CSV 내용을 확인하세요: {error}')
'''
app_path = Path('app.py')
app_path.write_text(app_code, encoding='utf-8')
print('생성 완료:', app_path.resolve())
print('터미널 실행: python -m streamlit run app.py')
````

### Q3. 이번 웹앱에서 MAE와 RMSE를 계산하지 않는 이유는 무엇인가요?

**답변:**  
예측 대상인 미래의 실제 대여량이 아직 관측되지 않아 예측 오차를 구할 수 없기 때문입니다. 향후 실제값이 확보되면 평가할 수 있습니다. 과거 기간의 성능 평가를 원하면 시간 순서에 맞춘 별도 Train/Test 분리가 필요합니다.
